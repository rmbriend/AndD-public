```python
%matplotlib inline

import numpy as np
import cv2 as cv
import matplotlib.pyplot as plt
from pathlib import Path
import os
```

```python
image_folder =Path("images")
template_folder = Path("template")

img_rgb = cv.imread('images/coca_retro_1.png')
#img_gray = cv.cvtColor(img_rgb, cv.COLOR_BGR2GRAY)
img_gray = cv.imread('images/coca_retro_1.png',0)
template = cv.imread('template/pattern.png',0)
inverted_template = 255 - template
#template = inverted_template
if img_gray.shape[0] < template.shape[0] or img_gray.shape[1] < template.shape[1]:
    print("Template is larger than the image. Resizing...")
    if img_gray.shape[0] < template.shape[0]:
        scale = img_gray.shape[0] / template.shape[0]
    else:
        scale = img_gray.shape[1] / template.shape[1]
    template = cv.resize(template, None, fx=scale, fy=scale)
w, h = template.shape[::-1]
#cv.imshow("imagen",img_gray)    
#cv.imshow("template",template)
res = cv.matchTemplate(img_gray,template,cv.TM_CCOEFF_NORMED)
# Encontramos los valores máximos y mínimos
min_val, max_val, min_loc, max_loc = cv.minMaxLoc(res)
top_left = max_loc
bottom_right = (top_left[0] + w, top_left[1] + h)
cv.rectangle(img_rgb,top_left, bottom_right, 255, 2)
# Fijamos un umbral de "coincidencia"
#for i in range(len(res)):
 #   if res[i] > 0.2:
  #      print (res[i])
print('Template: ',template.shape, 'image: ' ,img_gray.shape,'largo de resultados:',len(res),'forma de resultados:', res.shape,'ancho y alto de logo',w,h)

threshold = 0.2

loc = np.where( res >= threshold) 
print('coordenadas donde se supera el threshold:',loc)
#for pt in zip(*loc[::-1]): # el [::-1] es para invertir el orden de las coordenadas, necesitamos (x,y) y no (y,x)
#  cv.rectangle(img_rgb, pt, (pt[0] + w, pt[1] + h), (0,255,0), 2)

#   cv.imshow("imagen",img_rgb)    
#cv.imshow("template",template)
cv.imshow("imagen",img_rgb)    
cv.waitKey(0)
cv.destroyAllWindows()
```
