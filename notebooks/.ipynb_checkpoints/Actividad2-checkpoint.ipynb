```python
import pandas as pd
from pandas.api.types import CategoricalDtype
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats as st
from scipy.stats import describe
```

```python
crashes = pd.read_csv("https://raw.githubusercontent.com/rmbriend/AndD-public/refs/heads/master/datasets/aircrahesFullDataUpdated_2024.csv")
#df_csv = pd.read_csv("https://people.sc.fsu.edu/~jburkardt/data/csv/airtravel.csv")
```

```python
crashes.head()
```

```python
crashes.info()
```

```python
np.mean(crashes['Fatalities (air)'])
```

```python
np.median(crashes['Fatalities (air)'])
```

```python
fig, ax = plt.subplots(1, 2, figsize=(10, 5))

ax[0].hist(crashes['Fatalities (air)'],bins=50)   # fatalidades
ax[0].set_title('Fatalities (air)')
ax[1].hist(crashes['Aboard'],bins=50)  # posajeros
ax[1].set_title('Aboard')

plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 4))

sns.kdeplot(data=crashes, x='Fatalities (air)', fill=True, ax=axes[0])
sns.kdeplot(data=crashes, x='Aboard', fill=True, ax=axes[1])
axes[0].set_title('Fatalities')
axes[1].set_title('Aboard')

plt.tight_layout()
plt.show()
```

```python
sns.scatterplot(data=crashes, x='Aboard', y='Fatalities (air)', alpha=1)
plt.grid(ls='--')
plt.title('Relacion entre pasajeros y fatalidades')
plt.tight_layout()
plt.show()
```
